A **mention** = `@` followed by non-whitespace characters, e.g. `@POTUS`.
Regular expression: `@\\S+` (`\\S` = any non-space character, `+` = one or more).

```r
library(dplyr)
library(stringr)

load(url("https://dssoc.github.io/datasets/congress.RData"))
load(url("https://dssoc.github.io/datasets/senator_tweets.RData"))
```

## Count mentions in every tweet

```r
tweets <- senator_tweet_sample %>%
  mutate(mentions = str_count(text, "@\\S+"))

tweets %>% select(text, mentions) %>% head(5)
```

## (a) Correlation between number of mentions and number of favorites

```r
r <- cor(tweets$mentions, tweets$favorite_count)
cat("Correlation (mentions vs favorites):", round(r, 4), "\n")

cor.test(tweets$mentions, tweets$favorite_count)
```

## (b) Average number of mentions by gender (link tweets to senators through their Twitter handle)

```r
senators <- congress %>%
  inner_join(congress_contact, by = "bioguide_id") %>%
  mutate(twitter = tolower(twitter)) %>%
  select(twitter, gender)

tweets_gender <- tweets %>%
  mutate(screen_name = tolower(screen_name)) %>%
  inner_join(senators, by = c("screen_name" = "twitter"))

tweets_gender %>%
  group_by(gender) %>%
  summarise(avg_mentions = mean(mentions), tweets = n())
```
